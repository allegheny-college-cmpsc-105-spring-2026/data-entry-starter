# Data Entry Lab

---

Data extraction and entry refer to the systematic process of identifying/labeling and storing data from pre-existing material. See https://www.lenovo.com/us/en/glossary/data-entry/ for more information.

Data extraction and entry are a major part of what makes Data Exploration possible. The extraction processes can be thought of as pulling information or meta-information out of pre-existing material, and the entry part can be thought of as storing the extracted information. Without proper entry, data may not be in a useable format, and computer-based data explorations are not possible.

There are many curated datasets on the web (e.g. [Tableau Data](https://public.tableau.com/app/learn/sample-data); [Others](https://docs.google.com/spreadsheets/d/1uliFBPpcPtAoBczaGXCLgI00mMh7fUdTaObcTEnTtgw/edit?usp=sharing)), however it is likely that you will need to curate your own data at some point when conducting your own research.
